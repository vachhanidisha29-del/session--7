```python
import numpy as np
import pandas as pd

np.random.seed(42)


member_id = np.arange(1, 121)


plan = np.random.choice(
    ["Silver", "Gold", "Platinum"],
    120
)


monthly_fee_inr = np.where(
    plan == "Silver", 800,
    np.where(plan == "Gold", 1200, 1800)
)


gym_df = pd.DataFrame({
    "member_id": member_id,
    "plan": plan,
    "monthly_fee_inr": monthly_fee_inr,
    "months_paid": np.random.randint(1, 13, 120),
    "age": np.random.randint(18, 61, 120),
    "trainer_sessions": np.random.randint(0, 21, 120)
})


gym_df.to_csv("gym_members_raw.csv", index=False)

gym_df.head()
```

```python
gym_df["total_paid_inr"] = (
    gym_df["monthly_fee_inr"] * gym_df["months_paid"]
)

gym_df["gst_inr"] = (
    gym_df["total_paid_inr"] * 0.18
).round(2)


gym_df.head(5)
```

```python
gym_df["age_group"] = gym_df["age"].apply(
    lambda a: "Youth" if a < 25
    else ("Adult" if a <= 40 else "Senior")
)

gym_df[["member_id", "age", "age_group"]].head(10)
```

```python

currency_columns = [
    "monthly_fee_inr",
    "total_paid_inr",
    "gst_inr"
]


gym_display = gym_df[
    currency_columns
].copy()

gym_display = gym_display.map(
    lambda x: f"Rs {x:,.0f}"
)

gym_display.head(10)
```
